```python
import numpy as np
```

```python
A=np.arange(3,15).reshape((3,4))
print(A)
```

```python
print(A[1])
print(A[2][1])
print(A[2,1])
```

```python
for row in A:
    print(row)
```

```python
for column in A.T:
    print(column)
```

```python
print(A.flatten())
for item in A.flat:
    print(item)
```

```python
a=np.array([1,1,1])
b=np.array([2,2,2])
c=np.vstack((a,b)) #纵向合并
d=np.hstack((a,b)) #横向合并
print(c)
print(d)
print(c.shape)
print(d.shape)
```

```python
a=a[:,np.newaxis]
b=b[:,np.newaxis]
```

####np.newaxis用于给行或列加一个维度

np.reshape是用来改变数列形状的，但不改变内部数据，他能实现和newaxis同样的功能，但更强大

```python
x=np.concatenate((a,b,a,b),axis=1)#多个合并
print(x)
```

```python
print(np.split(A,2,axis=1))#分割数列，2表示分成几份
```

####np.array_split()用于数列的不等分割

np.split(A,2,axis=1) == np.hsplit(A,2)
np.split(A,2,axis=0) == np.vsplit(A,2)
