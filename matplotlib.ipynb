```python
x = np.linspace(-3,3,50)
y1 = 2*x+1
plt.figure()
plt.plot(x,y1,zorder=1)
#gca=提取坐标轴
ax = plt.gca()
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')
ax.xaxis.set_ticks_position('bottom')#下面这根线做x
ax.yaxis.set_ticks_position('left')#左边这根线做y
ax.spines['bottom'].set_position(('data',0))#让x轴锁定在y的0的位置
ax.spines['left'].set_position(('data',0))#定位方式除了data，还有outward,axes

#method1在图片上标点
x0 = 1
y0 = 2*x0 + 1
plt.scatter(x0,y0,s=50,color='b')#标记一个点
plt.plot([x0,x0],[y0,0],'k--',lw=2.5)#给这个点加入纵虚线


#method2在图上标注函数
plt.annotate(r'$2x+1=%s$' %y0,xy=(x0,y0),xycoords='data',xytext=(+30,-30),textcoords='offset points',
            fontsize=16,arrowprops=dict(arrowstyle='->',connectionstyle='arc3,rad=.2'))

#method3在图上标注文字
plt.text(-3.7,3,r'$This\ is\ the\ some\ text.\ \mu\ \sigma_i\ \alpha_t$',
        fontdict={'size':16,'color':'r'})

#能见度
plt.draw()
for label in ax.get_xticklabels() + ax.get_yticklabels():
    label.set_fontsize(12)
    label.set_bbox(dict(facecolor='white',edgecolor='None',alpha=0.7))
plt.show()
```

```python
n=1024
X=np.random.normal(0,1,n)
Y=np.random.normal(0,1,n)
T=np.arctan2(X,Y)# for color value
plt.scatter(X,Y,s=75,c=T,alpha=0.5)#alpha为透明度
plt.xlim((-1.5,1.5))
plt.ylim((-1.5,1.5))
plt.show()
```

```python
#柱状图
n=12
X1=np.arange(n)
Y1=(1-X1/float(n))*np.random.uniform(0.5,1.0,n)
Y2=(1-X1/float(n))*np.random.uniform(0.5,1.0,n)
plt.bar(X1,+Y1,facecolor='#9999ff',edgecolor='white')
plt.bar(X1,-Y2,facecolor='#ff9999',edgecolor='white')

for x,y in zip(X1,Y1):
    plt.text(x,y+0.05,'%.2f' % y,ha='center',va='bottom')
for x,y in zip(X1,Y2):
    plt.text(x,-y-0.05,'%.2f' % y,ha='center',va='top')
    #ha:horizontal alignment
plt.show()
```

```python
#等高线
def f(x,y):
    return(1-x/2 + x**5 + y**3)*np.exp(-x**2-y**2)
n=256
x = np.linspace(-3,3,n)
y = np.linspace(-3,3,n)
X,Y=np.meshgrid(x,y)
plt.contourf(X,Y,f(X,Y),8,alpha=0.75,cmap=plt.cm.hot)
#plt.contourf for filling contours
#X,Y and value for (X,Y)point
C = plt.contour(X,Y,f(X,Y),8,colors='black',linewidth=.5)
plt.clabel(C,inline=False,fontsize=10)
```

```python
#3D图像
from mpl_toolkits.mplot3d import Axes3D
%matplotlib ipympl
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
X=np.arange(-4,4,0.25)
Y=np.arange(-4,4,0.25)
X,Y=np.meshgrid(X,Y)
R = np.sqrt(X**2+Y**2)
Z = np.sin(R)
ax.plot_surface(X,Y,Z,rstride=1,cstride=1,cmap='rainbow',edgecolor='black')
ax.contourf(X,Y,Z,zdir='z',offset=-2,cmap='rainbow')#投影出等高线

```

```python
#动画展示数据animiation
from matplotlib import animation
fig,ax = plt.subplots()
x = np.arange(0,2*np.pi,0.01)
line, = ax.plot(x,np.sin(x))#因为返回值是列表，所以要打逗号，选择列表第一位的值
def animate(i):#这里的i指的是frame的值
    line.set_ydata(np.sin(x+i/10))#更新的y值
    return line,

def init():#定义起始值
    line.set_ydata(np.sin(x))
    return line,

ani = animation.FuncAnimation(fig=fig,func=animate,frames=100,init_func=init,interval=20,blit=False)
#frame指长度一百帧；init_func指第一帧什么样；interval指更新的频率默认毫秒，blit（false）更新所有点，（true）更新变化点
```
