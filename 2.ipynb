```python
A=np.arange(2,14).reshape((3,4))
print(A)
```

```python
print(np.argmin(A))  
print(np.argmax(A)) 
print(np.cumsum(A))
print(np.diff(A))
print(np.nonzero(A))
print(A.T)
```

###最小值的索引：np.argmin(A)；最大值的索引：np.argmax(A)


np.mean(A)=A.mean()输出平均值//mean(A,axis=0),指的是对行进行求均值，若axis等于1则对列进行求均值


np.cumsum(A)逐个累加


np.diff(A)依次两个相减（后减前）


np.nonzero(A)输出所有非零数的索引


np.sort(A)对所有元素进行排序


np.transpose(A) == A.T求矩阵的转置


np.clip(A,5,9)让A中所有小于5的数都等于5，所有大于9的数都等于9
